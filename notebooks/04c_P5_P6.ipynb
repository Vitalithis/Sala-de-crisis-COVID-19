## P5 · Grupos de países: cuántos, qué los caracteriza y dónde está Chile

## P6 · Ómicron: pronóstico con corte en dic-2021 frente a lo real; días de anticipación de la alerta

```python
spec = importlib.util.spec_from_file_location('mod04', str(RAIZ / 'src' / '04_modelos_y_alerta.py'))
mod = importlib.util.module_from_spec(spec); spec.loader.exec_module(mod)
pred = mod.pronosticar('Chile', '2021-12-28', horizonte=14, dw_path=str(RAIZ / 'data' / 'dw_covid.duckdb'))
real = mod.obtener_serie_territorio('Chile', dw_path=str(RAIZ / 'data' / 'dw_covid.duckdb'))
comp = pred.merge(real[['fecha', 'mm7_casos']], on='fecha')
display(comp)
bt = pd.read_csv(RAIZ / 'data' / 'clean' / 'backtesting_chile.csv')
display(bt[bt['fecha_corte'] == '2021-12-28'])
# Al corte 2021-12-28 la alerta aún estaba inactiva; la alerta temprana se activó el 2022-01-03 (41 días antes del pico de ómicron del 2022-02-13).
```
